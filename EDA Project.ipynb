```python
# Delete Null Values

df3.dropna()
df3.dropna(how="all")
```

```python
# Data Visualition
# Bar Chart

plt.bar(df3["department"],df3["salary"],color="red")
plt.title("Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```

```python
# Line Chart

plt.plot(df3["age"],df3["salary"],marker='o',color="red")
plt.title("Age by Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# Box Plot

plt.boxplot(df3["salary"])
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

```python
# Pie Chart

plt.pie(df3["id"],labels=df3["department"],autopct="%1.0f%%")
plt.title("Employee Distribution by Dept")
plt.show()
```
